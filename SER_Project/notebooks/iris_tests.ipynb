## Tests using sklearn iris dataset ##

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np

data = load_breast_cancer()
cancer = pd.DataFrame(data.data, columns=data.feature_names)
target = data.target
print(cancer.head())

import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
```

```python
# Reference regression test
from sklearn.linear_model import LogisticRegression
from minilearn.preprocessing import train_test_split

X_train, X_test, y_train, y_test = train_test_split(cancer, data.target, test_size=0.2, random_state=42)
model = LogisticRegression(max_iter=100, random_state=42)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)
print(f"Test Accuracy: {accuracy:.2f}")
```

```python
from minilearn.classifiers import LogisticRegression

model = LogisticRegression(n_iterations=100, learning_rate=0.01)
model.fit(X_train, y_train)
accuracy = model.score(X_test, y_test)
print(f"MiniLearn Test Accuracy: {accuracy:.2f}")
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train, y_train)
knn_accuracy = knn_model.score(X_test, y_test)
print(f"KNN Test Accuracy: {knn_accuracy:.2f}")
```

```python
from minilearn.classifiers import KNN
knn_model = KNN(n_neighbors=3)
knn_model.fit(X_train, y_train)
knn_accuracy = knn_model.score(X_test, y_test)
print(f"MiniLearn KNN Test Accuracy: {knn_accuracy:.2f}")
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
nb_accuracy = nb_model.score(X_test, y_test)
print(f"Naive Bayes Test Accuracy: {nb_accuracy:.2f}")
```

```python
from minilearn.classifiers import GaussianNaiveBayes
nb_model = GaussianNaiveBayes()
nb_model.fit(X_train, y_train)
nb_accuracy = nb_model.score(X_test, y_test)
print(f"MiniLearn Naive Bayes Test Accuracy: {nb_accuracy:.2f}")
```

```python
from sklearn.svm import SVC
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train, y_train)
svm_accuracy = svm_model.score(X_test, y_test)
print(f"SVM Test Accuracy: {svm_accuracy:.2f}")
```

```python
from minilearn.classifiers import SVM
svm_model = SVM(kernel='linear', n_iterations=100, learning_rate=0.01, lambda_param=0.01)
svm_model.fit(X_train, y_train)
svm_accuracy = svm_model.score(X_test, y_test)
print(f"MiniLearn SVM Test Accuracy: {svm_accuracy:.2f}")
```

```python
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_accuracy = dt_model.score(X_test, y_test)
print(f"Decision Tree Test Accuracy: {dt_accuracy:.2f}")
```

```python
from minilearn.classifiers import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(criterion="entropy", max_depth=4)
dt_model.fit(X_train, y_train)
dt_accuracy = dt_model.score(X_test, y_test)
print(f"MiniLearn Decision Tree Test Accuracy: {dt_accuracy:.2f}")
```
